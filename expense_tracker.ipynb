```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create expense data
data = {
    "Item": ["Food", "Travel", "Shopping", "Food", "Travel"],
    "Amount": [200, 100, 500, 150, 80]
}

# Step 2: Create DataFrame
df = pd.DataFrame(data)

# Step 3: Show data
print("Expense Data:\n")
print(df)

# Step 4: Total expense
total = df["Amount"].sum()
print("\nTotal Expense:", total)

# Step 5: Category-wise sum
category_total = df.groupby("Item")["Amount"].sum()

print("\nCategory-wise Expense:\n")
print(category_total)

# Step 6: Graph
category_total.plot(kind='bar')

plt.title("Expense by Category")
plt.xlabel("Category")
plt.ylabel("Amount")

plt.show()
```
